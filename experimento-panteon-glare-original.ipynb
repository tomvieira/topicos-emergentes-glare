```python
# %%
# Experimento: NTv3 (modelo base, congelado) + classificador leve
# + controles + análise de acerto por identidade ao treino (MMseqs2)
# Entrada: arquivos .fa por classe em  <DATA_ROOT>/<dataset>/Train/  e  .../Test/
#          (um arquivo por classe, ex.: TIR.fa, LTR.fa, Non-TE.fa)
# Requisito para a seção 7: MMseqs2 instalado e acessível no PATH (comando `mmseqs`)
import os
import re
import time
import shutil
import hashlib
import subprocess
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, normalize
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import confusion_matrix, classification_report, f1_score
import pandas as pd
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModelForMaskedLM, AutoConfig

# %%
# =============================================
# CONFIGURAÇÃO
# =============================================
torch.manual_seed(42)
np.random.seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Usando o dispositivo: {device}")

dataset = 'panteon-original'
DATA_ROOT = './Data'                             # ajuste para sua estrutura
TRAIN_DIR = os.path.join(DATA_ROOT, dataset, 'Train')
VAL_DIR = os.path.join(DATA_ROOT, dataset, 'Validation') # <-- NOVA PASTA DE VALIDAÇÃO
TEST_DIR = os.path.join(DATA_ROOT, dataset, 'Test')

MODEL_ID = "InstaDeepAI/NTv3_8M_pre"             # versões: 8M, 100M, 650M (sufixo _pre)
PESOS_ALEATORIOS = False                         # True = controle sem pré-treino
CAMADA = -1                                      # índice do hidden_state a usar

WIN = 4096                                       # tamanho da janela (nt)
STRIDE = 2048                                    # passo entre janelas (50% de sobreposição)
MAX_WINDOWS = 8                                  # máx. de janelas por sequência (controla custo)
BATCH_WIN = 16                                   # janelas por lote na extração

USAR_PESO_DE_CLASSE = False                      # True se o seu dataset for desbalanceado
ORDEM_EXIBICAO = None                            # ex.: ["LTR", "LINE", "CLASS II", "Non-TE"] ou None

# --- Análise de identidade (MMseqs2) ---
RODAR_MMSEQS = False                              # False = reaproveita o hits_*.tsv já gerado
MMSEQS_THREADS = 4
LIMIAR_ID = 0.8                                  # teste "honesto": só sequências com identidade < limiar
PASTA_ANALISE = './Data/analise_ntv3'

CACHE_DIR = './Data/embeddings_ntv3'
os.makedirs(CACHE_DIR, exist_ok=True)
os.makedirs(PASTA_ANALISE, exist_ok=True)
TAG = MODEL_ID.split('/')[-1] + ('_rand' if PESOS_ALEATORIOS else '') + f'_L{CAMADA}'

# %%
# =============================================
# 1. LEITURA DOS FASTA (um arquivo por classe)
# =============================================
EXTS = ('.fa', '.fasta', '.fna')


def ler_fasta(path):
    ids, seqs = [], []
    cur_id, cur = None, []
    with open(path) as fh:
        for line in fh:
            line = line.strip()
            if not line:
                continue
            if line.startswith('>'):
                if cur_id is not None:
                    ids.append(cur_id)
                    seqs.append(''.join(cur))
                cur_id, cur = line[1:].split()[0], []
            else:
                cur.append(line)
    if cur_id is not None:
        ids.append(cur_id)
        seqs.append(''.join(cur))
    return ids, seqs


def limpar(seq):
    seq = seq.upper().replace('U', 'T')
    return re.sub(r'[^ACGTN]', 'N', seq)


def carregar_split(pasta, classes=None):
    arquivos = sorted(f for f in os.listdir(pasta) if f.lower().endswith(EXTS))
    nomes = [os.path.splitext(f)[0] for f in arquivos]
    if classes is None:
        classes = nomes
    else:
        extras = set(nomes) - set(classes)
        if extras:
            raise ValueError(f"Classes no {pasta} ausentes no treino: {extras}")
    ids, seqs, labels = [], [], []
    for f, nome in zip(arquivos, nomes):
        i, s = ler_fasta(os.path.join(pasta, f))
        ids += i
        seqs += [limpar(x) for x in s]
        labels += [classes.index(nome)] * len(s)
    return ids, seqs, np.array(labels, dtype=np.int64), classes


ids_tr, seqs_tr, y_tr, classes = carregar_split(TRAIN_DIR)
ids_val, seqs_val, y_val, _ = carregar_split(VAL_DIR, classes)
ids_te, seqs_te, y_te, _ = carregar_split(TEST_DIR, classes)
num_classes = len(classes)

print("Classes:", classes)
print("Treino:", {c: int((y_tr == i).sum()) for i, c in enumerate(classes)})
print("Validação:", {c: int((y_val == i).sum()) for i, c in enumerate(classes)})
print("Teste :", {c: int((y_te == i).sum()) for i, c in enumerate(classes)})

# %%
# =============================================
# 2. NTv3: janelas + embeddings (modelo congelado)
# =============================================
def carregar_modelo():
    tok = AutoTokenizer.from_pretrained(MODEL_ID, trust_remote_code=True)
    if PESOS_ALEATORIOS:
        cfg = AutoConfig.from_pretrained(MODEL_ID, trust_remote_code=True)
        model = AutoModelForMaskedLM.from_config(cfg, trust_remote_code=True)
    else:
        model = AutoModelForMaskedLM.from_pretrained(MODEL_ID, trust_remote_code=True)
    return tok, model.to(device).eval()


def janelas(seq):
    if len(seq) <= WIN:
        return [seq]
    starts = list(range(0, len(seq) - WIN + 1, STRIDE))
    if starts[-1] + WIN < len(seq):
        starts.append(len(seq) - WIN)
    if len(starts) > MAX_WINDOWS:
        idx = np.linspace(0, len(starts) - 1, MAX_WINDOWS).round().astype(int)
        starts = [starts[i] for i in idx]
    return [seq[s:s + WIN] for s in starts]


@torch.no_grad()
def extrair_embeddings(seqs, tok, model):
    itens = [(i, w) for i, s in enumerate(seqs) for w in janelas(s)]
    ordem = sorted(range(len(itens)), key=lambda k: len(itens[k][1]))  # reduz padding
    soma, cont = None, np.zeros(len(seqs))
    t0 = time.time()
    for b in range(0, len(ordem), BATCH_WIN):
        idx = ordem[b:b + BATCH_WIN]
        enc = tok([itens[k][1] for k in idx], add_special_tokens=False, padding=True,
                  pad_to_multiple_of=128, return_tensors='pt')
        enc = {k: v.to(device) for k, v in enc.items() if k in ('input_ids', 'attention_mask')}
        out = model(**enc, output_hidden_states=True)
        h = out.hidden_states[CAMADA]
        pad_id = tok.pad_token_id
        if pad_id is None:
            mask = torch.ones_like(enc['input_ids']).unsqueeze(-1).float()
        else:
            mask = (enc['input_ids'] != pad_id).unsqueeze(-1).float()
        if mask.shape[1] != h.shape[1]:   # camadas intermediárias podem ter resolução menor
            mask = nn.functional.adaptive_max_pool1d(mask.transpose(1, 2), h.shape[1]).transpose(1, 2)
        vec = ((h.float() * mask).sum(1) / mask.sum(1).clamp(min=1)).cpu().numpy()
        if soma is None:
            soma = np.zeros((len(seqs), vec.shape[1]), dtype=np.float32)
        for j, k in enumerate(idx):
            soma[itens[k][0]] += vec[j]
            cont[itens[k][0]] += 1
        if (b // BATCH_WIN) % 50 == 0:
            print(f"  {b}/{len(ordem)} janelas ({time.time() - t0:.0f}s)")
    return soma / cont[:, None]


def obter_embeddings(nome_split, seqs):
    path = os.path.join(CACHE_DIR, f"{TAG}_{dataset}_{nome_split}.npy")
    if os.path.exists(path):
        print(f"Carregando cache: {path}")
        return np.load(path)
    tok, model = carregar_modelo()
    emb = extrair_embeddings(seqs, tok, model)
    np.save(path, emb)
    del model
    torch.cuda.empty_cache()
    return emb


# %%
# Teste rápido (rode antes do passo completo): confere formato e tempo
tok_t, model_t = carregar_modelo()
t0 = time.time()
e = extrair_embeddings(seqs_tr[:4], tok_t, model_t)
print("Formato dos embeddings:", e.shape, f"| {time.time() - t0:.1f}s para 4 sequências")
del model_t
torch.cuda.empty_cache()

# %%
X_tr = obter_embeddings('train', seqs_tr)
X_val = obter_embeddings('val', seqs_val)
X_te = obter_embeddings('test', seqs_te)
print("Treino:", X_tr.shape, "| Validação:", X_val.shape, "| Teste:", X_te.shape)

# %%
# =============================================
# 3. Preparação dos dados para o treino
# =============================================
scaler = StandardScaler().fit(X_tr)
X_tr = scaler.transform(X_tr)
X_val = scaler.transform(X_val)
X_te_s = scaler.transform(X_te)


class EmbDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.long)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, i):
        return self.X[i], self.y[i]


train_loader = DataLoader(EmbDataset(X_tr, y_tr), batch_size=32, shuffle=True)
val_loader = DataLoader(EmbDataset(X_val, y_val), batch_size=32)
test_loader = DataLoader(EmbDataset(X_te_s, y_te), batch_size=32)   # sem shuffle: ordem = seqs_te

# %%
# =============================================
# 4. Classificador leve sobre os embeddings
# =============================================
class MLPHead(nn.Module):
    def __init__(self, dim, hidden_dim=256, num_classes=10):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(dim, hidden_dim), nn.ReLU(), nn.Dropout(0.4),
            nn.Linear(hidden_dim, hidden_dim // 2), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(hidden_dim // 2, num_classes))

    def forward(self, x):
        return self.net(x)


model = MLPHead(X_tr.shape[1], 256, num_classes).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-4)

if USAR_PESO_DE_CLASSE:
    cont = np.bincount(y_tr, minlength=num_classes).astype(np.float32)
    pesos = torch.tensor(cont.sum() / (num_classes * np.maximum(cont, 1)), device=device)
    criterion = nn.CrossEntropyLoss(weight=pesos)
else:
    criterion = nn.CrossEntropyLoss()


def acc_batch(out, y):
    return (out.argmax(1) == y).float().mean().item()


num_epochs, patience = 100, 20
best_val_loss, sem_melhora = float('inf'), 0
best_model_path = f"ntv3_head_{TAG}_{dataset}.pt"
train_losses, val_losses, train_accs, val_accs = [], [], [], []

for epoch in range(num_epochs):
    model.train()
    tl = ta = 0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        optimizer.zero_grad()
        out = model(x)
        loss = criterion(out, y)
        loss.backward()
        optimizer.step()
        tl += loss.item()
        ta += acc_batch(out, y)
    tl /= len(train_loader)
    ta /= len(train_loader)

    model.eval()
    vl = va = 0
    with torch.no_grad():
        for x, y in val_loader:
            x, y = x.to(device), y.to(device)
            out = model(x)
            vl += criterion(out, y).item()
            va += acc_batch(out, y)
    vl /= len(val_loader)
    va /= len(val_loader)

    train_losses.append(tl); val_losses.append(vl)
    train_accs.append(ta); val_accs.append(va)
    print(f"Epoch {epoch+1}: Train Loss = {tl:.4f}, Train Acc = {ta:.4f}, "
          f"Val Loss = {vl:.4f}, Val Acc = {va:.4f}")

    if vl < best_val_loss:
        best_val_loss, sem_melhora = vl, 0
        torch.save(model.state_dict(), best_model_path)
    else:
        sem_melhora += 1
        if sem_melhora >= patience:
            print(f"Early stopping após {epoch+1} épocas")
            break

# %%
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(train_losses, label='Train Loss'); plt.plot(val_losses, label='Val Loss')
plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.title('Loss Curves')
plt.subplot(1, 2, 2)
plt.plot(train_accs, label='Train Acc'); plt.plot(val_accs, label='Val Acc')
plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.title('Accuracy Curves')
plt.tight_layout()
plt.show()

# %%
# =============================================
# 5. Avaliação no teste
# =============================================
model.load_state_dict(torch.load(best_model_path))
model.eval()
all_preds, all_labels = [], []
with torch.no_grad():
    for x, y in test_loader:
        all_preds.extend(model(x.to(device)).argmax(1).cpu().numpy())
        all_labels.extend(y.numpy())
all_preds, all_labels = np.array(all_preds), np.array(all_labels)
assert (all_labels == y_te).all(), "Ordem do teste não confere com seqs_te"

# Ordem de exibição (mesma para matriz e métricas)
if ORDEM_EXIBICAO is not None:
    ordem_idx = [classes.index(c) for c in ORDEM_EXIBICAO]
else:
    ordem_idx = list(range(num_classes))
nomes_exib = [classes[i] for i in ordem_idx]

cm = confusion_matrix(all_labels, all_preds, labels=ordem_idx)
print(cm)

fig, ax = plt.subplots(figsize=(10, 10))
im = ax.imshow(cm, interpolation='nearest', cmap='Blues')
ax.set_xlabel('Predicted label', fontsize=18)
ax.set_ylabel('True label', fontsize=18)
ax.set_xticks(np.arange(len(nomes_exib)))
ax.set_yticks(np.arange(len(nomes_exib)))
ax.set_xticklabels(nomes_exib, fontsize=14)
ax.set_yticklabels(nomes_exib, fontsize=14)
plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")
for t in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(j, t, f"{int(cm[t, j])}", ha="center", va="center",
                color="white" if t == j else "black", fontsize=18)
cbar = ax.figure.colorbar(im, ax=ax)
cbar.ax.tick_params(labelsize=14)
plt.tight_layout()
plt.show()

# %%
metrics = classification_report(all_labels, all_preds, labels=ordem_idx,
                                target_names=nomes_exib, output_dict=True, zero_division=0)
metrics_df = pd.DataFrame(metrics).transpose()


def accuracy_by_class(cm):
    accs = []
    for i in range(cm.shape[0]):
        TP = cm[i, i]
        FP = cm[:, i].sum() - TP
        FN = cm[i, :].sum() - TP
        TN = cm.sum() - TP - FP - FN
        total = TP + TN + FP + FN
        accs.append((TP + TN) / total if total > 0 else 0.0)
    return accs


def specificity(cm):
    out = []
    for i in range(cm.shape[0]):
        FP = cm[:, i].sum() - cm[i, i]
        TN = cm.sum() - cm[i, :].sum() - cm[:, i].sum() + cm[i, i]
        out.append(TN / (TN + FP) if (TN + FP) > 0 else 0.0)
    return out


def format_by_class(vals, names):
    return [f'{names[i]}: {v:.3f}' for i, v in enumerate(vals)]


acc_by_class = accuracy_by_class(cm)
spec_by_class = specificity(cm)
print("Acurácia por classe:", format_by_class(acc_by_class, nomes_exib))
print(f"Acurácia Macro: {np.mean(acc_by_class):.3f}")
print("Especificidade por classe:", format_by_class(spec_by_class, nomes_exib))
print(f"Especificidade Macro: {np.mean(spec_by_class):.3f}")

df_new = metrics_df.loc[nomes_exib].copy()
df_new['specificity'] = spec_by_class
df_new['acc'] = acc_by_class
df_new = df_new.reindex(columns=['acc', 'precision', 'recall', 'f1-score', 'specificity', 'support'])
df_new = pd.concat([df_new, pd.DataFrame([df_new.mean(numeric_only=True)], index=['Mean'])])
print(df_new)

print(f"\nF1 macro   : {f1_score(all_labels, all_preds, average='macro'):.4f}")
print(f"F1 ponderado: {f1_score(all_labels, all_preds, average='weighted'):.4f}")
df_new.to_csv(f"resultados_{TAG}_{dataset}.csv")

# %%
# =============================================
# 6. Controles (mesmo treino/teste)
#    a) regressão logística nos embeddings do NTv3
#    b) k-meros (k=5) + regressão logística
#    c) rótulos embaralhados (deve ficar perto do acaso; se não, há bug)
# =============================================
lr = LogisticRegression(max_iter=2000).fit(X_tr, y_tr)
pred_lr = lr.predict(X_te_s)
print(f"[NTv3 + LogReg]   F1 macro: {f1_score(y_te, pred_lr, average='macro'):.4f} | "
      f"F1 pond.: {f1_score(y_te, pred_lr, average='weighted'):.4f}")

vec = CountVectorizer(analyzer='char', ngram_range=(5, 5), lowercase=False)
Xk_all = normalize(vec.fit_transform([seqs_tr[i] for i in range(len(seqs_tr))]), norm='l1')
Xk_te = normalize(vec.transform(seqs_te), norm='l1')
sc_k = StandardScaler(with_mean=False).fit(Xk_all)
lr_k = LogisticRegression(max_iter=2000).fit(sc_k.transform(Xk_all), y_tr)
pred_kmer = lr_k.predict(sc_k.transform(Xk_te))
print(f"[k-mer + LogReg]  F1 macro: {f1_score(y_te, pred_kmer, average='macro'):.4f} | "
      f"F1 pond.: {f1_score(y_te, pred_kmer, average='weighted'):.4f}")

rng = np.random.default_rng(0)
lr_0 = LogisticRegression(max_iter=2000).fit(X_tr, rng.permutation(y_tr))
pred_0 = lr_0.predict(X_te_s)
print(f"[Rótulo embaralhado] F1 macro: {f1_score(y_te, pred_0, average='macro'):.4f} "
      f"(esperado: perto do acaso ~ {1 / num_classes:.2f})")

# %%
# =============================================
# 7. ANÁLISE DE VAZAMENTO: acerto por identidade ao treino (MMseqs2)
#    Pergunta: o modelo acerta porque aprendeu a classificar ou porque o
#    teste tem sequências quase iguais às do treino?
# =============================================
# 7.0 Duplicatas exatas e IDs repetidos entre treino e teste
md5 = lambda s: hashlib.md5(s.encode()).hexdigest()
set_tr = {md5(s) for s in seqs_tr}
dup_exata = np.array([md5(s) in set_tr for s in seqs_te])
print(f"Sequências do teste idênticas a alguma do treino: {dup_exata.sum()} de {len(seqs_te)}")
print(f"IDs repetidos entre treino e teste: {len(set(ids_tr) & set(ids_te))}")

# 7.1 Busca MMseqs2 (nucleotídeos, duas fitas). Usamos IDs numéricos (te_i / tr_j)
#     para evitar problemas com os caracteres especiais dos cabeçalhos originais.
hits_path = os.path.join(PASTA_ANALISE, f'hits_{dataset}.tsv')


def escrever_fasta(path, seqs, prefixo):
    with open(path, 'w') as f:
        for i, s in enumerate(seqs):
            f.write(f">{prefixo}{i}\n{s}\n")


mmseqs_ok = True
if RODAR_MMSEQS or not os.path.exists(hits_path):
    fa_te = os.path.join(PASTA_ANALISE, 'test.fa')
    fa_tr = os.path.join(PASTA_ANALISE, 'train.fa')
    escrever_fasta(fa_te, seqs_te, 'te_')
    escrever_fasta(fa_tr, seqs_tr, 'tr_')
    tmp = os.path.join(PASTA_ANALISE, 'tmp')
    shutil.rmtree(tmp, ignore_errors=True)
    cmd = ['mmseqs', 'easy-search', fa_te, fa_tr, hits_path, tmp,
           '--search-type', '3', '--strand', '2', '-s', '7.5',
           '--min-seq-id', '0.3', '-c', '0.5',
           '--threads', str(MMSEQS_THREADS),
           '--format-output', 'query,target,fident,qcov']
    try:
        subprocess.run(cmd, check=True)
    except FileNotFoundError:
        print("MMseqs2 não encontrado no PATH. Instale (conda install -c bioconda mmseqs2) e rode de novo.")
        mmseqs_ok = False

# %%
if mmseqs_ok:
    hits = pd.read_csv(hits_path, sep='\t', names=['q', 't', 'fident', 'qcov'])
    hits = hits[hits.qcov >= 0.5]                                   # descarta alinhamentos curtos
    hits = hits.sort_values('fident', ascending=False).drop_duplicates('q')   # melhor hit por consulta
    qi = hits.q.str.replace('te_', '', regex=False).astype(int).values
    ti = hits.t.str.replace('tr_', '', regex=False).astype(int).values

    max_id = np.zeros(len(seqs_te))                 # 0 = sem hit
    max_id[qi] = hits.fident.values
    rotulo_hit = np.full(len(seqs_te), -1)          # classe do melhor hit no treino
    rotulo_hit[qi] = y_tr[ti]

    faixas = [-0.01, 0.0, 0.7, 0.8, 0.9, 1.0]
    nomes_faixa = ['sem hit', '<=0.7', '0.7-0.8', '0.8-0.9', '>0.9']
    faixa = pd.cut(max_id, faixas, labels=nomes_faixa)

    metodos = {
        'NTv3 + MLP': all_preds,
        'NTv3 + LogReg': pred_lr,
        'k-mer + LogReg': pred_kmer,
    }

    tabela = pd.DataFrame({'n': pd.Series(faixa).value_counts().reindex(nomes_faixa)})
    for nome, p in metodos.items():
        tabela[nome] = pd.Series(p == y_te).groupby(faixa, observed=False).mean().reindex(nomes_faixa)
    # Baseline de homologia: classe do melhor hit (só onde existe hit)
    acerto_hit = np.where(max_id > 0, rotulo_hit == y_te, np.nan)
    tabela['Melhor hit (homologia)'] = pd.Series(acerto_hit).groupby(faixa, observed=False).mean().reindex(nomes_faixa)

    print("\nAcerto por faixa de identidade máxima ao treino:")
    print(tabela.round(3))
    tabela.to_csv(os.path.join(PASTA_ANALISE, f'acerto_por_faixa_{TAG}_{dataset}.csv'))

    ax = tabela.drop(columns='n').plot.bar(figsize=(10, 5), rot=0)
    ax.set_xlabel('Identidade máxima ao treino (MMseqs2)')
    ax.set_ylabel('Acurácia')
    ax.set_ylim(0, 1.05)
    ax.set_title(f'{dataset}: acerto por faixa de identidade ({TAG})')
    plt.tight_layout()
    plt.show()

    # 7.2 Teste "honesto": só sequências sem parente próximo no treino
    m = max_id < LIMIAR_ID
    print(f"\nSequências com identidade < {LIMIAR_ID}: {m.sum()} de {len(y_te)}")
    print("Por classe (restantes):",
          {c: int(((y_te == i) & m).sum()) for i, c in enumerate(classes)})
    labs = np.unique(y_te[m])
    print(f"\n{'Método':<20} {'F1 macro (todos)':>18} {'F1 macro (<' + str(LIMIAR_ID) + ')':>18}")
    for nome, p in metodos.items():
        f_all = f1_score(y_te, p, average='macro')
        f_sub = f1_score(y_te[m], p[m], average='macro', labels=labs, zero_division=0)
        print(f"{nome:<20} {f_all:>18.4f} {f_sub:>18.4f}")

    # 7.3 Salva as predições com a identidade de cada sequência
    pd.DataFrame({
        'id': ids_te, 'real': y_te, 'pred_mlp': all_preds, 'pred_lr': pred_lr,
        'pred_kmer': pred_kmer, 'max_id': max_id, 'faixa': faixa.astype(str),
        'classe_real': [classes[i] for i in y_te],
        'classe_pred_mlp': [classes[i] for i in all_preds],
    }).to_csv(os.path.join(PASTA_ANALISE, f'predicoes_{TAG}_{dataset}.csv'), index=False)
    print(f"\nArquivos salvos em {PASTA_ANALISE}")
```

```python
# %% [markdown]
# # GLARE (receita fiel) aplicado ao Dataset3
#
# Reimplementação do pipeline de treino/avaliação do GLARE (`glare/model.py`, `glare/data.py`, `glare/train.py`, config `..._rc_augment_best_hpo_final.yaml`) lendo **seu** formato: `Data/<Dataset>/Train/*.fa` e `Test/*.fa` (um FASTA por classe).
#
# **Única mudança intencional:** `base_model` = `NTv3_8M_pre` (o GLARE usa `NTv3_650M_pre`) e o conjunto de classes = as do D3. Todo o resto segue o código do GLARE.
#
# | Item | Seu notebook antigo | GLARE (esta versão) |
# |---|---|---|
# | Head | hidden encolhido p/ 256 (256→256→128) | sempre 512→256, dropout 0.2232 |
# | Pooling | mean (ok) | mean mascarado (igual) |
# | Avaliação | só os primeiros 4096 nt | janelas deslizantes (50% overlap), média dos logits por sequência |
# | Seleção de checkpoint | menor val loss, patience 5 | maior F1 ponderado na validação, patience 8 |
# | Loss | CE + LS 0.05 | CE ponderada `sqrt_inv` + LS 0.0465 |
# | Otimizador | AdamW 5e-5 / 3e-4, wd 0.01, sem scheduler | AdamW 2.53e-5 / 2.89e-4, wd 0.0597, cosseno + 5% warmup |
# | Batch | 8 | 8 × acumulação 4, `drop_last=True` |
# | Precisão | fp16 sem GradScaler | bf16 autocast |
# | Val split | 20%, seed 42 | 10% estratificado, seed 7 |
# | Augment | RC antes do crop | crop aleatório e depois RC (p=0.5) |
# | Tokenização | `add_special_tokens=False`, limpeza IUPAC→N | chamada idêntica à do GLARE; só `.upper()` |

# %%
!pip install -q scikit-learn matplotlib
# Se o carregamento do NTv3 falhar, o GLARE pina estas versões:
# !pip install -q "transformers>=4.40,<4.45" "tokenizers>=0.19,<0.20"

# %%
import os, re, time, glob, math, random, inspect, importlib, sys
import numpy as np
import torch, torch.nn as nn
from torch.utils.data import Dataset, DataLoader
# train_test_split removido (divisão física)
from sklearn.metrics import classification_report, confusion_matrix, f1_score, accuracy_score
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoConfig, AutoModel, AutoModelForMaskedLM, get_cosine_schedule_with_warmup

os.environ["TOKENIZERS_PARALLELISM"] = "false"

# ------------------------- DADOS (único bloco "seu") -------------------------
DATASET   = "Dataset3"
DATA_ROOT = "./Data"
TRAIN_DIR = os.path.join(DATA_ROOT, DATASET, "Train")
VAL_DIR   = os.path.join(DATA_ROOT, DATASET, "Validation") # <-- NOVA PASTA ADICIONADA
TEST_DIR  = os.path.join(DATA_ROOT, DATASET, "Test")
EXTS = (".fa", ".fasta", ".fna")

# --------------- CONFIG = espelho do YAML best_hpo_final do GLARE ---------------
CFG = dict(
    base_model="InstaDeepAI/NTv3_8M_pre",   # GLARE: InstaDeepAI/NTv3_650M_pre
    max_length=4096,
    hidden_dim=512,
    head_dropout=0.22320323256729352,
    batch_size=8,
    max_epochs=10,
    patience=8,
    lambda_sf=1.0,
    backbone_lr=2.53032090218936e-05,
    head_lr=0.00028944361832243895,
    weight_decay=0.059703505248527425,
    warmup_fraction=0.05,
    # val_fraction removido do CFG, pois a divisão é física
    seed=7,
    label_smoothing=0.04654581174590394,
    class_weight_scheme="sqrt_inv",
    grad_accum=4,            # YAML=4. O paper rodou em 4 GPUs (batch efetivo 128); em 1 GPU use 16 p/ igualar.
    rc_augment=True,
    overlap=0.5,
)
CKPT_PATH = "glare_d3_best.pt"

random.seed(CFG["seed"]); np.random.seed(CFG["seed"]); torch.manual_seed(CFG["seed"])
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
USE_BF16 = device.type == "cuda" and torch.cuda.is_bf16_supported()
print("device:", device, "| bf16 autocast:", USE_BF16)
if device.type == "cuda" and not USE_BF16:
    print("AVISO: GPU sem bf16 (ex.: T4). Rodando em fp32 — único desvio imposto pelo hardware.")

# %%
# ============================ LEITURA DOS FASTA ============================
def ler_fasta(path):
    ids, seqs, cur_id, cur = [], [], None, []
    with open(path, errors="replace") as fh:
        for line in fh:
            line = line.strip()
            if not line: continue
            if line.startswith(">"):
                if cur_id is not None: ids.append(cur_id); seqs.append("".join(cur))
                cur_id, cur = line[1:], []
            else:
                cur.append(line)
    if cur_id is not None: ids.append(cur_id); seqs.append("".join(cur))
    return ids, seqs

def carregar_split(pasta, classes=None):
    arquivos = sorted(f for f in os.listdir(pasta) if f.lower().endswith(EXTS))
    nomes = [os.path.splitext(f)[0] for f in arquivos]
    if classes is None: classes = nomes
    recs = []
    for f, nome in zip(arquivos, nomes):
        _, seqs = ler_fasta(os.path.join(pasta, f))
        for s in seqs:   # GLARE: apenas .upper() (parse_panteon_fasta)
            recs.append({"sequence": s.upper(), "label": classes.index(nome)})
    return recs, classes

# Carregamento físico direto das 3 pastas
train_recs, classes = carregar_split(TRAIN_DIR)
val_recs, _ = carregar_split(VAL_DIR, classes)   # <-- CARREGA DA PASTA VALIDATION
test_recs, _ = carregar_split(TEST_DIR, classes)

NUM_CLASSES = len(classes)
L = CFG["max_length"]

print(f"classes ({NUM_CLASSES}): {classes}")
print(f"train={len(train_recs)} val={len(val_recs)} test={len(test_recs)}")
print(f"seqs > {L} nt: train {np.mean([len(r['sequence'])>L for r in train_recs]):.1%} | "
      f"test {np.mean([len(r['sequence'])>L for r in test_recs]):.1%}  (nessas, a janela deslizante faz diferença)")

# %%
# ============================ TOKENIZER + DATASETS ============================
tokenizer = AutoTokenizer.from_pretrained(CFG["base_model"], trust_remote_code=True)
PAD_ID = tokenizer.pad_token_id if tokenizer.pad_token_id is not None else 0
RC_MAP = str.maketrans("ACGTacgt", "TGCAtgca")   # igual ao glare/data.py

def encode(seq):
    # chamada idêntica à de TEDataset/SlidingWindowDataset do GLARE
    enc = tokenizer(seq, return_tensors="pt", padding="max_length", truncation=True,
                    max_length=L, pad_to_multiple_of=128)
    ids = enc["input_ids"].squeeze(0)
    mask = enc["attention_mask"].squeeze(0) if "attention_mask" in enc else (ids != PAD_ID).long()
    return ids, mask

class TEDataset(Dataset):                      # treino (glare.data.TEDataset, augment=True)
    def __init__(self, recs, rc_augment): self.recs, self.rc = recs, rc_augment
    def __len__(self): return len(self.recs)
    def __getitem__(self, i):
        r = self.recs[i]; s = r["sequence"]
        if len(s) > L:                                   # 1) crop aleatório
            st = random.randint(0, len(s) - L); s = s[st:st+L]
        if self.rc and random.random() < 0.5:            # 2) reverse-complement
            s = s.translate(RC_MAP)[::-1]
        ids, mask = encode(s)
        return {"input_ids": ids, "attention_mask": mask,
                "label": torch.tensor(r["label"], dtype=torch.long)}

def sequence_windows(seq, max_length, overlap=0.5):   # cópia de glare.data.sequence_windows
    if len(seq) <= max_length: return [seq]
    stride = max(1, int(max_length * (1.0 - overlap)))
    out, start = [], 0
    while start < len(seq):
        end = min(start + max_length, len(seq)); out.append(seq[start:end])
        if end == len(seq): break
        start += stride
    return out

class SlidingWindowDataset(Dataset):           # val/teste (glare.data.SlidingWindowDataset)
    def __init__(self, recs, overlap):
        self.w = [(i, win, r["label"]) for i, r in enumerate(recs)
                  for win in sequence_windows(r["sequence"], L, overlap)]
        self.n_sequences = len(recs)
    def __len__(self): return len(self.w)
    def __getitem__(self, k):
        i, win, y = self.w[k]; ids, mask = encode(win)
        return {"input_ids": ids, "attention_mask": mask,
                "label": torch.tensor(y), "seq_idx": torch.tensor(i)}

NW = min(4, os.cpu_count() or 4)
def mk(ds, shuffle=False, drop_last=False):
    return DataLoader(ds, batch_size=CFG["batch_size"], shuffle=shuffle, drop_last=drop_last,
                      num_workers=NW, pin_memory=True, persistent_workers=NW > 0,
                      prefetch_factor=2 if NW > 0 else None)

train_loader = mk(TEDataset(train_recs, CFG["rc_augment"]), shuffle=True, drop_last=True)
val_ds  = SlidingWindowDataset(val_recs,  CFG["overlap"]); val_loader  = mk(val_ds)
test_ds = SlidingWindowDataset(test_recs, CFG["overlap"]); test_loader = mk(test_ds)
print(f"janelas: val={len(val_ds)} test={len(test_ds)}")

# %%
# ============================ MODELO (glare/model.py) ============================
class GLAREHead(nn.Module):
    # order_head omitido: no config final lambda_order=0 (sem efeito no treino/predição) e o D3 não tem 'order'.
    def __init__(self, input_dim, hidden_dim, num_classes, dropout):
        super().__init__()
        self.trunk = nn.Sequential(
            nn.Linear(input_dim, hidden_dim), nn.LayerNorm(hidden_dim), nn.GELU(), nn.Dropout(dropout),
            nn.Linear(hidden_dim, hidden_dim // 2), nn.LayerNorm(hidden_dim // 2), nn.GELU(), nn.Dropout(dropout))
        self.superfamily_head = nn.Linear(hidden_dim // 2, num_classes)
    def forward(self, x): return self.superfamily_head(self.trunk(x))

def load_backbone(name):
    config = AutoConfig.from_pretrained(name, trust_remote_code=True)
    if device.type == "cuda" and not USE_BF16:        # só em GPU sem bf16
        for k in ("stem_compute_dtype", "down_convolution_compute_dtype",
                  "up_convolution_compute_dtype", "core_compute_dtype"):
            setattr(config, k, "float32")
    try:                                              # 1) AutoModel (backbone sem cabeça de MLM)
        return AutoModel.from_pretrained(name, config=config, trust_remote_code=True)
    except Exception as e1:
        print("AutoModel falhou:", repr(e1)[:200])
    try:                                              # 2) caminho do GLARE: NTv3PreTrained via módulo em cache
        hits = glob.glob(os.path.expanduser("~/.cache/huggingface/modules/transformers_modules/**/modeling_ntv3_pretrained.py"),
                         recursive=True)
        sys.path.insert(0, os.path.dirname(hits[0]))
        mod = importlib.import_module("modeling_ntv3_pretrained")
        return mod.NTv3PreTrained.from_pretrained(name, config=config, trust_remote_code=True)
    except Exception as e2:
        print("NTv3PreTrained falhou:", repr(e2)[:200])
    print("AVISO: usando AutoModelForMaskedLM (como no seu notebook antigo); confira o hidden size abaixo.")
    return AutoModelForMaskedLM.from_pretrained(name, config=config, trust_remote_code=True)

class GLAREModel(nn.Module):
    def __init__(self, name, num_classes):
        super().__init__()
        self.backbone = load_backbone(name)
        for p in self.backbone.parameters(): p.requires_grad = True       # full fine-tuning (use_lora=False)
        cfg = self.backbone.config
        hs = getattr(cfg, "hidden_size", None) or getattr(cfg, "d_model", None)
        if hs is None: hs = self._probe_hidden()
        self.hidden_size = int(hs)
        self.head = GLAREHead(self.hidden_size, CFG["hidden_dim"], num_classes, CFG["head_dropout"])
        sig = inspect.signature(self.backbone.forward).parameters
        self._kw = {k for k in ("input_ids", "attention_mask", "output_hidden_states") if k in sig}

    @torch.no_grad()
    def _probe_hidden(self):
        ids = torch.full((1, 128), PAD_ID if PAD_ID else 1, dtype=torch.long)
        return self._last_hidden(self.backbone(input_ids=ids, output_hidden_states=True)).shape[-1]

    @staticmethod
    def _last_hidden(o):                      # mesma ordem de preferência do GLARE
        if getattr(o, "hidden_states", None) is not None: return o.hidden_states[-1]
        if getattr(o, "last_hidden_state", None) is not None: return o.last_hidden_state
        return o[0]

    def forward(self, input_ids, attention_mask):
        kw = dict(input_ids=input_ids, attention_mask=attention_mask, output_hidden_states=True)
        h = self._last_hidden(self.backbone(**{k: v for k, v in kw.items() if k in self._kw}))
        m = attention_mask.unsqueeze(-1).float()                           # mean pool mascarado
        pooled = (h * m).sum(1) / m.sum(1).clamp(min=1)
        return self.head(pooled)

model = GLAREModel(CFG["base_model"], NUM_CLASSES).to(device)
n_tr = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"hidden_size do backbone = {model.hidden_size} | parâmetros treináveis = {n_tr/1e6:.1f}M")

# %%
# ============================ LOSS, OTIMIZADOR, SCHEDULER ============================
def compute_class_weights(labels, num_classes, scheme="sqrt_inv"):   # cópia de glare.train
    counts = np.bincount(labels, minlength=num_classes).astype(float)
    present = counts > 0; w = np.zeros(num_classes)
    if scheme == "none": w[present] = 1.0
    elif scheme == "inv": w[present] = 1.0 / counts[present]; w[present] /= w[present].mean()
    else: w[present] = 1.0 / np.sqrt(counts[present]); w[present] /= w[present].mean()
    return torch.tensor(w, dtype=torch.float32)

cw = compute_class_weights([r["label"] for r in train_recs], NUM_CLASSES, CFG["class_weight_scheme"]).to(device)
criterion = nn.CrossEntropyLoss(weight=cw, label_smoothing=CFG["label_smoothing"])

optimizer = torch.optim.AdamW(
    [{"params": [p for p in model.backbone.parameters() if p.requires_grad], "lr": CFG["backbone_lr"]},
     {"params": list(model.head.parameters()), "lr": CFG["head_lr"]}],
    weight_decay=CFG["weight_decay"])

total_steps  = (len(train_loader) // CFG["grad_accum"]) * CFG["max_epochs"]
warmup_steps = int(total_steps * CFG["warmup_fraction"])
scheduler = get_cosine_schedule_with_warmup(optimizer, warmup_steps, total_steps)
print(f"optimizer steps totais={total_steps} warmup={warmup_steps}")

# %%
# ============================ TREINO / AVALIAÇÃO ============================
def autocast_ctx():
    return torch.autocast(device_type="cuda", dtype=torch.bfloat16, enabled=USE_BF16)

def train_epoch():
    model.train(); tot, nb = 0.0, 0; n = len(train_loader); ga = CFG["grad_accum"]
    optimizer.zero_grad(set_to_none=True)
    for i, b in enumerate(train_loader):
        ids, am, y = b["input_ids"].to(device), b["attention_mask"].to(device), b["label"].to(device)
        last = ((i + 1) % ga == 0) or (i + 1) == n
        with autocast_ctx():
            loss = CFG["lambda_sf"] * criterion(model(ids, am), y)
        (loss / ga).backward()
        tot += loss.item(); nb += 1
        if last:
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step(); scheduler.step(); optimizer.zero_grad(set_to_none=True)
    return tot / max(nb, 1)

@torch.no_grad()
def evaluate(loader, n_seq, with_loss=True):
    model.eval()
    logit_sum = torch.zeros(n_seq, NUM_CLASSES, device=device)
    cnt = torch.zeros(n_seq, device=device); lab = torch.full((n_seq,), -1, dtype=torch.long, device=device)
    tot, nb = 0.0, 0
    for b in loader:
        ids, am, y, si = (b["input_ids"].to(device), b["attention_mask"].to(device),
                          b["label"].to(device), b["seq_idx"].to(device))
        with autocast_ctx():
            lg = model(ids, am)
            if with_loss: tot += (CFG["lambda_sf"] * criterion(lg, y)).item(); nb += 1
        logit_sum.index_add_(0, si, lg.float()); cnt.index_add_(0, si, torch.ones_like(si, dtype=torch.float))
        lab[si] = y
    pred = (logit_sum / cnt.clamp(min=1).unsqueeze(1)).argmax(-1).cpu().numpy()   # média dos logits por sequência
    true = lab.cpu().numpy()
    return tot / max(nb, 1), pred, true

best_f1, no_imp = -1.0, 0
print("Iniciando fine-tuning completo (receita GLARE)…")
for ep in range(1, CFG["max_epochs"] + 1):
    t0 = time.time(); tr_loss = train_epoch()
    v_loss, vp, vt = evaluate(val_loader, val_ds.n_sequences)
    v_f1 = f1_score(vt, vp, average="weighted", zero_division=0)
    print(f"Epoch {ep}/{CFG['max_epochs']} | {time.time()-t0:.0f}s | train_loss={tr_loss:.4f} | "
          f"val_loss={v_loss:.4f} | val_wF1={v_f1:.4f} | val_acc={accuracy_score(vt, vp):.4f}")
    if v_f1 > best_f1:
        best_f1, no_imp = v_f1, 0; torch.save(model.state_dict(), CKPT_PATH); print("  ✓ novo melhor — checkpoint salvo")
    else:
        no_imp += 1
        if no_imp >= CFG["patience"]: print("Early stopping."); break

# %%
# ============================ TESTE (melhor checkpoint por val F1) ============================
model.load_state_dict(torch.load(CKPT_PATH, map_location=device)); model.to(device)
_, tp, tt = evaluate(test_loader, test_ds.n_sequences, with_loss=False)

print(classification_report(tt, tp, labels=list(range(NUM_CLASSES)), target_names=classes, digits=4, zero_division=0))
print(f"F1 ponderado: {f1_score(tt, tp, average='weighted', zero_division=0):.4f} | "
      f"F1 macro: {f1_score(tt, tp, average='macro', zero_division=0):.4f} | acc: {accuracy_score(tt, tp):.4f}")

cm = confusion_matrix(tt, tp, labels=list(range(NUM_CLASSES)))
fig, ax = plt.subplots(figsize=(10, 10)); im = ax.imshow(cm, cmap="Blues")
ax.set_xlabel("Rótulo Previsto", fontsize=18); ax.set_ylabel("Rótulo Verdadeiro", fontsize=18)
ax.set_xticks(range(NUM_CLASSES)); ax.set_yticks(range(NUM_CLASSES))
ax.set_xticklabels(classes, fontsize=12, rotation=45, ha="right", rotation_mode="anchor"); ax.set_yticklabels(classes, fontsize=12)
for i in range(NUM_CLASSES):
    for j in range(NUM_CLASSES):
        ax.text(j, i, int(cm[i, j]), ha="center", va="center", color="white" if i == j else "black", fontsize=11)
fig.colorbar(im, ax=ax); plt.tight_layout(); plt.show()
```
